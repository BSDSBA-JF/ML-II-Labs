```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
smote_df = pd.read_csv("spreadsheets/summary_smote_experiments_statistics.csv")

rate_values = smote_df.iloc[3].values.astype(float)  
value_values = smote_df.iloc[5].values.astype(float)  
techniques = smote_df.columns.tolist()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

x_pos = np.arange(len(techniques)) 
bar_width = 0.35 
offset = bar_width / 2 

bars1 = ax1.bar(x_pos - offset, rate_values, bar_width, color='#7c31e4', label='Fraud Capture Rate')
ax1.set_xlabel('Technique')
ax1.set_ylabel('Fraud Capture Rate')
ax1.set_ylim(0.6, 1.0)

# Add labels to the rate bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 0.005,
             f'{height:.2f}',
             ha='center', va='bottom', fontsize=9)

ax2 = ax1.twinx() 
bars2 = ax2.bar(x_pos + offset, value_values, bar_width, color='#157497', label='Absolute Net Value')
ax2.set_ylabel('Absolute Net Value (€)') 
ax2.set_ylim(3500, 4300) 

# Add labels to the value bars
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 20, 
             f'{height:.0f}', 
             ha='center', va='bottom', fontsize=9)

ax1.set_title('Fraud Capture Rate and Absolute Net Value by Technique', fontsize=14)
ax1.set_xticks(x_pos) 
ax1.set_xticklabels(techniques, rotation=45, ha="right")

lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left') 
plt.tight_layout() 
plt.show()
```
