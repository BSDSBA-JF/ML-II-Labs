```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("spreadsheets/summary_smote_experiments_statistics.csv")

rate_values = df.iloc[3].values.astype(float)  
value_values = df.iloc[5].values.astype(float)  
techniques = df.columns.tolist()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

x_pos = np.arange(len(techniques))
bar_width = 0.8 # Width of the bars


bars1 = axes[0].bar(x_pos, rate_values, bar_width, color='#7c31e4') 
axes[0].set_title('Fraud Capture Rate by Technique', fontsize=14)
axes[0].set_ylabel('Fraud Capture Rate (0.0 to 1.0)')
axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(techniques, rotation=45, ha="right")
axes[0].set_ylim(.6, 1.0)


for bar in bars1:
    height = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2., height + 0.02,
            f'{height:.2f}',
            ha='center', va='bottom', fontsize=10)

bars2 = axes[1].bar(x_pos, value_values, bar_width, color='#157497') 
axes[1].set_title('Absolute Net Value by Technique', fontsize=14)
axes[1].set_ylabel('Absolute Net Value ($)')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(techniques, rotation=45, ha="right")
axes[1].set_ylim(3500, 4300)

# Add data labels
for bar in bars2:
    height = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2., height + 5, 
            f'{height:.2f}',
            ha='center', va='bottom', fontsize=10)


plt.tight_layout(rect=[0, 0.03, 1, 0.95]) 
plt.savefig(f'figures/{fig_count:02d}_barchart.png', dpi=300, bbox_inches='tight')
fig_count += 1
plt.show()
```
