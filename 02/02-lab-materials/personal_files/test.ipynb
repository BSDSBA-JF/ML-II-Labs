```python
import pandas as pd
import joblib
import numpy as np
from sklearn.metrics import classification_report
```

```python
old_model = joblib.load("model.joblib")
```

```python
df = pd.read_csv("historical.csv")

old_pred = old_model.predict(df)
```

```python
df["risk_level"] 
df['risk_level'] = np.where(df['risk_level'] == 'At Risk', 1, 0)

real = df["risk_level"] 
```

```python
report = classification_report(real,old_pred)
print(report)
```
