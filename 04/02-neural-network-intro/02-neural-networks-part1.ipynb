## Backpropagation: This changes everything!

The introduction of the Perceptron created the very first hype around AI around the late 1950s, but quickly met its demise due to its limitations.

Rumelhart (lead), Hinton, and Williams in their paper [Learning representations by backpropagating errors](https://www.nature.com/articles/323533a0) back in 1986. The idea is simple: make the activation function continuous and backpropagate the errors from the front to the back. 

*Note: Of course this is an oversimplification of their work, but I want to illustrate that even simple changes and ideas could spark revolutionary changes. **Stay creative.***


A family of continuous functions that would reliably approximate the step function are sigmoid functions.

<div class="row" align="center">
<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/8/88/Logistic-curve.svg/500px-Logistic-curve.svg.png" width="30%"></img>

There are many members of the sigmoid family, but the they generally look like this:

\begin{equation*}
f(x) = \frac{A}{1 + e^{-k(x-x_o)}},
\end{equation*}

where $A$ is the maximum value, $x_o$ is the location of the midpoint, and $k$ is the steepness of the curve. Essentially, sigmoids are defined by two horizontal asymptotes. This is the main reason why they are often used for learning curves, activations functions, and other growth models that exhibit an approximately exponential increase that saturates at a certain point.

### Single output neural network...from scratch

Let's now create a neural network from scratch. Let's follow the architecture below and we'll learn a few more concepts along the way.

<figure style="text-align:center">
<img src="figures/basic-network.png" width="40%">
</figure>

Let's first try to understand how this neural network works mathematically.

**Step 1. Forward Propagation**
The functional mapping from inputs $(x_1, x_2)$ to the output $\Psi_{NN}$ is given by

$$\Psi_{NN} = g(x_1, x_2) = g(w_1f_1(x_1), w_2f_2(x_2)),$$

where $w_1$ and $w_2$ are the learnable weights, and $f_1, f_2$, and $g$ are the activation functions. 

**Step 2. Define the error**

If you want your system a mapping from $\Psi_{NN}$ to $\Psi_{\text{actual}}$, then we need a way to let it know when it's getting things wrong. In this case, we need to define an error function $E$. 

$$E = \frac{1}{2}(\Psi_{NN} - \Psi_{\text{actual}})^2$$

Having a defined error function allows us to update $w_1$ and $w_2$. Note that the $1/2$ is only placed for mathematical convenience.

**Step 3. Implement Gradient Descent**

We then need an update rule, so that for each iteration our weights get updated. This is done using gradient descent.

$$w_1^{t + 1} = w_1^t - \gamma\frac{\partial E}{\partial w_1^t}$$
$$w_2^{t + 1} = w_2^t - \gamma\frac{\partial E}{\partial w_2^t}$$

For this derivation, let's assume that we will be using the sigmoid we defined earlier for all activation functions. 

$$w_1^{t + 1} = w_1^t - \gamma(\Psi_{NN} - \Psi_{\text{actual}})g'(x_1, x_2)f_1(x_1)$$
$$w_2^{t + 1} = w_2^t - \gamma(\Psi_{NN} - \Psi_{\text{actual}})g'(x_1, x_2)f_2(x_2)$$

Note that $g'(x) = g(x)(1-g(x))$, we then get

$$w_1^{t + 1} = w_1^t - \gamma(\Psi_{NN} - \Psi_{\text{actual}})g(\cdot)(1 - g(\cdot))f_1(x_1)$$
$$w_2^{t + 1} = w_2^t - \gamma(\Psi_{NN} - \Psi_{\text{actual}})g(\cdot)(1 - g(\cdot))f_2(x_2)$$

where $\gamma$ is called the learning rate. And that's it! Time to code.

```python
# Define the activation function
def activation_function(x, deriv=False):
    sigmoid =  1 / (1 + np.exp(-x))
    if deriv:
        return sigmoid * (1 - sigmoid)
    return sigmoid
```

```python
# Define our constants
gamma = 0.005
n_input_nodes = 2
n_output_nodes = 1


# Define inputs
# each row is a sample and col a feature
X = np.array(
    [
        [-1, -1],
        [-1, 1],
        [1, -1],
        [1, 1]
    ]
)

# Change the targets based on the chosen Psi column
y = np.array([[0, 1, 1, 0]]).T

# initialize weights
# have the initial weights have mean 0 and std = 1
# this is 2 by 1
w0 = np.random.normal(loc=0, scale=1, size=(n_input_nodes, n_output_nodes)) 
n_iters = int(1e4)

print(f"Initial weights:\n{w0}")
print("=" * 50)

# train the network
for i in range(n_iters):
    layer0 = X
    layer1 = activation_function(np.dot(layer0, w0))

    error = layer1 - y
    if i % 1000 == 0:
        print(f"Iter: {i} --- Error: {np.mean(np.abs(error)):.4f}")
    
    layer1_delta = error * activation_function(np.dot(layer0, w0), deriv=True)
    w0 -= gamma * np.dot(layer0.T, layer1_delta)


y_preds = np.rint(layer1).flatten()

print("=" * 50)
print("Raw Output after training:\n")
print(layer1)
print("Learned weights:\n")
print(w0)
print("-" * 50)
print(f"Thresholded result at 0.5:{y_preds}")
print(f"Accuracy: {accuracy_score(y, y_preds) * 100:.2f}%")
```

```python
# Define the activation function
def activation_function(x, deriv=False):
    sigmoid =  1 / (1 + np.exp(-x))
    if deriv:
        return sigmoid * (1 - sigmoid)
    return sigmoid
```

```python
# Define our constants

# gamma = learning rate
gamma1 = 0.01
gamma2 = 0.01

n_input_nodes = 2
n_hidden_nodes = 2
n_output_nodes = 1

# Define inputs

X = np.array(
    [
        [-1, -1],
        [-1, 1],
        [1, -1],
        [1, 1]
    ]
)
# bias = [1, 1, 1, 1] # These are not learnable in this case
# X = np.insert(X, 0, bias, axis=1)
X = np.hstack([X, np.ones((X.shape[0], 1))])  # Add bias column


y = np.array([[0, 1, 1, 1]]).T

# initialize weights
# have the initial weights have mean 0 and std = 1
w1 = 2 * np.random.random((n_input_nodes + 1, n_hidden_nodes)) - 1  
w2 = 2 * np.random.random((n_hidden_nodes + 1, n_output_nodes)) - 1 
n_iters = int(1e4)

print(f"Initial weights:\n")
print(f"{w1=}\n")
print(f"{w2=}\n")
print("=" * 50)

# train the network
for i in range(n_iters):
    layer0 = X

    layer1_input = np.dot(layer0, w1)
    layer1 = activation_function(layer1_input)
    layer1_b = np.hstack([layer1, np.ones((layer1.shape[0], 1))])  

    layer2_input = np.dot(layer1_b, w2)
    layer2 = activation_function(layer2_input)

    error = layer2 - y
    if i % 1000 == 0:
        print(f"Iter {i} --- Error: {np.mean(np.abs(error)):.2f}")
    
    layer2_delta = error * activation_function(layer2_input, deriv=True)
    layer1_delta = layer2_delta.dot(w2.T) * activation_function(layer1_b, deriv=True)
    layer1_delta = layer1_delta[:, :-1]

    w2 -= gamma2 * layer1_b.T.dot(layer2_delta)
    w1 -= gamma1 * layer0.T.dot(layer1_delta)


    
y_preds = np.rint(layer2).flatten()
print("=" * 50)
print("Raw Output after training:\n")
print(layer2)
print("Learned weights:\n")
print(f"{w1=}\n")
print(f"{w2=}\n")
print("-" * 50)
print(f"Thresholded result at 0.5:{y_preds}")
print(f"Accuracy: {accuracy_score(y, y_preds) * 100:.2f}%")
```

### How do we make predictions on our trained network?

So far we've discussed about training a neural network, but we haven't gotten to the point of actually making predictions (inferences) on new data. To make our discussion simpler, let's just use the same inputs. 

We've seen that in the process of training, we're conducting backpropagation. However for inference tasks, we only need to do the forward propagation. We simply take our trained weights and pass them to our inputs. We'll implement the `forward` and `predict` functions for the forward propagation and inference, respectively.

```python
# Define the activation function
def activation_function(x, deriv=False):
    sigmoid =  1 / (1 + np.exp(-x))
    if deriv:
        return sigmoid * (1 - sigmoid)
    return sigmoid
```

```python
def forward(X, w1, w2):

    # add a constant bias column
    X_with_bias = np.hstack([X, np.ones((X.shape[0], 1))])

    # alternatively, you can use the @ operator 
    # for more convenience
    # Ex: input_to_hidden = X_with_bias @ w1
    input_to_hidden = np.dot(
        X_with_bias, w1
    )
    
    hidden = activation_function(input_to_hidden)
    hidden_with_bias = np.hstack(
        [hidden, np.ones((hidden.shape[0], 1))]
    )
    hidden_to_out = np.dot(hidden_with_bias, w2)
    return activation_function(hidden_to_out)

def predict(X, w1, w2, thresh=0.5):
    output_prob = forward(X, w1, w2)
    return np.rint(layer2).flatten()
```
