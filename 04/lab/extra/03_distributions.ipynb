```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
char_df = pd.read_csv('data/cleaned_node.csv')

char_df
```

Basically, we see here that the elements don't really affect if you are going to be on or off the field. However, what type of role a character is (DPS, Support, or Survivability) and if they are on or off field are not independent. This is because if a character is a support or for survivability, they are more likely to be off the field.

```python
# Plot out the number of on v off fielders per element
fielders = ["On-Field", "Off-Field", ]
element_counts = char_df.groupby("Element")[fielders].sum()


element_counts.plot(kind="bar", stacked=True)


plt.ylabel("Count")
plt.title("On-Field vs Off-Field per Element")

max_count = element_counts.sum(axis=1).max()
plt.yticks(range(0, max_count + 2, 3))

plt.xticks(rotation=0)
plt.legend(title="Status")

plt.legend(title="Status", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
roles = ["DPS", "Support", "Survivability"]

roles_count = pd.DataFrame()
for role in roles:
    counts = char_df.groupby(role)[fielders].sum()
    roles_count[role] = counts.T[1]

roles_count.T.plot(kind='bar', stacked=True)

plt.xticks(rotation=0)
plt.legend(title="Status")

plt.legend(title="Status", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
