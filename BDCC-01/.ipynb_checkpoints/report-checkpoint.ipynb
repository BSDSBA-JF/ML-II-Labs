```python
import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
folder_path = "/mnt/data/public/food-com-recipes/RAW_recipes.csv"
df = pd.read_csv(folder_path)
df["nutrition"] = df["nutrition"].apply(json.loads)
```

```python
# Get the nutritional values
nutrition_cols = [
    "calories",
    "total_fat_PDV",
    "sugar_PDV",
    "sodium_PDV",
    "protein_PDV",
    "saturated_fat_PDV",
    "carbohydrates_PDV",
]

df_nutritions = df["nutrition"].apply(pd.Series)
df_nutritions.columns = nutrition_cols
df_nutritions.head()
```

## Healthy Food Classification Criteria

The nutritional profile of food items was classified using a rule-based system aligned with dietary guidelines for a moderately active adult male consuming 2,400 kcal per day. A food item is considered **healthy** only if it satisfies **all** of the following criteria:

### 1. Caloric Appropriateness (Energy Control)
- The total caloric content must be **less than or equal to 800 calories per meal**.

**Rule:**
calories ≤ 800



### 2. Beneficial Nutrient Density
- The food must be a significant source of at least one key beneficial nutrient, defined as providing **20% or more of the Daily Value (%DV)**.

**Rule:**
protein_PDV ≥ 20 OR carbohydrates_PDV ≥ 20


*(Carbohydrate %DV is assumed to be beneficial since we suppose these represent complex carbohydrates or fiber.)*

### 3. Limiting Nutrient Control
- The food must be low in nutrients associated with chronic disease risk. According to FDA guidance, a nutrient is considered low if it contributes **5% DV or less**.

**Rules:**
saturated_fat_PDV ≤ 5
sugar_PDV ≤ 5
sodium_PDV ≤ 5



---

### **Final Logical Expression (Combined Rule)**

A food item is classified as **healthy** if and only if:

(calories ≤ 800)
AND
(protein_PDV ≥ 20 OR carbohydrates_PDV ≥ 20)
AND
(saturated_fat_PDV ≤ 5)
AND
(sugar_PDV ≤ 5)
AND
(sodium_PDV ≤ 5)


Any item not meeting all of these conditions is classified as **not healthy**.


```python
# Get the healthy foods
is_healthy = (
    (df_nutritions["calories"] <= 800)
    & (
        (df_nutritions["protein_PDV"] >= 20)
        | (df_nutritions["carbohydrates_PDV"] >= 20)
    )
    & (df_nutritions["saturated_fat_PDV"] <= 5)
    & (df_nutritions["sugar_PDV"] <= 5)
    & (df_nutritions["sodium_PDV"] <= 5)
)

healthy_foods = df_nutritions[is_healthy].index
df_healthy = df.loc[
    healthy_foods, ["name", "id", "minutes", "tags", "nutrition", "ingredients"]
]

# Convert the tags into 
df_healthy['tags'] = df_healthy['tags'].apply(
    lambda x: [item.strip("'\"") for item in x.strip("[]").split(", ")]
)
df_healthy['ingredients'] = df_healthy['ingredients'].apply(
    lambda x: [item.strip("'\"") for item in x.strip("[]").split(", ")]
)
```

```python
df_healthy.to_json('data/healthy_foods.json', orient='records')
```

```python
df_nutritions.loc[healthy_foods].to_json('data/nutrition.json', orient='records')
```
