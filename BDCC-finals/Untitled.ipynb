```python
import os
import pandas as pd

base_dir = "/mnt/data/public/insideairbnb/data.insideairbnb.com/united-states"
states = os.listdir(base_dir)
state_cities = {
    state: [city for city in os.listdir(os.path.join(base_dir, state))]
    for state in states
}
```

```python
all_data = []
start_year = 2015


for state, cities in state_cities.items():
    for city in cities:

        #print(state, city)

        df_data_list = []
        city_base_dir = os.path.join(base_dir, state, city)

        # Filter folders that start with a valid year
        year_folders = [
            f for f in os.listdir(city_base_dir)
            if f[:4].isdigit() and int(f[:4]) >= start_year
        ]

        for year_folder in sorted(year_folders):

            data_filepath = os.path.join(
                city_base_dir,
                year_folder,
                "visualisations",
                "listings.csv"
            )
            all_data.append(data_filepath)
            data_filepath = os.path.join(
                city_base_dir,
                year_folder,
                "data",
                "listings.csv.gz"
            )
            all_data.append(data_filepath)
            data_filepath = os.path.join(
                city_base_dir,
                year_folder,
                "visualisations",
                "neighbourhoods.geojson")
            if os.path.exists(data_filepath):
                all_data.append(data_filepath)
            
            data_filepath = os.path.join(
                city_base_dir,
                year_folder,
                "visualisations",
                "neighbourhoods.csv")
            if os.path.exists(data_filepath):
                all_data.append(data_filepath)
```

```python
total_mem_mb = 0

total_size_bytes = sum(os.path.getsize(f) for f in all_data)
total_size_mb = total_size_bytes / (1024 * 1024)
total_size_gb = total_size_mb / 1024  # convert MB to GB

print(f"Total size on disk: {total_size_gb:.2f} GB")
```
